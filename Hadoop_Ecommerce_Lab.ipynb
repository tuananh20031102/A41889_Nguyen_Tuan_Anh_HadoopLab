```python
%%writefile /content/mapper_revenue.py
import sys
import csv

reader = csv.DictReader(sys.stdin)

for row in reader:
    if row["status"] == "SUCCESS":
        category = row["category"]
        amount = int(row["amount"])
        print(f"{category}\t{amount}")
```

```python
%%writefile /content/reducer_revenue.py
import sys

current_category = None
total = 0

for line in sys.stdin:
    line = line.strip()

    if not line:
        continue

    category, amount = line.split("\t")
    amount = int(amount)

    if current_category == category:
        total += amount
    else:
        if current_category is not None:
            print(f"{current_category}\t{total}")

        current_category = category
        total = amount

if current_category is not None:
    print(f"{current_category}\t{total}")
```

### 1. Kết quả doanh thu theo nhóm sản phẩm

| Hạng | Category | Tổng doanh thu |

### 2. Nhận xét kết quả

Dữ liệu đầu vào là file `orders_2026_09.csv`, chứa thông tin các giao dịch thương mại điện tử trong tháng 09/2026.  
Bài toán chỉ tính doanh thu đối với những giao dịch có trạng thái `SUCCESS`.  
Mapper đọc từng dòng CSV và phát sinh cặp key-value theo dạng `category → amount`.  
Sau bước Shuffle/Sort, các bản ghi có cùng category được gom lại với nhau.  
Reducer cộng tổng amount của từng category và ghi kết quả vào file `part-00000`.  
Kết quả cho thấy Electronics có doanh thu cao nhất với 3,703,769,311, tiếp theo là Home với 410,315,065 và Fashion với 218,597,763.  
Kết quả này có thể hỗ trợ doanh nghiệp ưu tiên quản lý tồn kho, chương trình khuyến mại và nguồn lực giao hàng cho các nhóm sản phẩm có doanh thu cao.

### 3. Hạn chế của xử lý batch

Hadoop MapReduce phù hợp với xử lý dữ liệu lớn theo lô, tuy nhiên kết quả không được cập nhật tức thời. Khi dữ liệu mới phát sinh, cần chạy lại job để tạo kết quả mới, vì vậy mô hình này không phù hợp với các bài toán yêu cầu phản hồi thời gian thực.
